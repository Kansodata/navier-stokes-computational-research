# Simulation Analysis

This notebook inspects the metrics emitted by a completed run under `outputs/baseline_random/metrics.csv`.

```python
from pathlib import Path
import csv
import matplotlib.pyplot as plt

metrics_path = Path('outputs/baseline_random/metrics.csv')
if not metrics_path.exists():
    raise FileNotFoundError('Run the baseline experiment first: nsim --config configs/baseline_random.json')

with metrics_path.open('r', encoding='utf-8') as handle:
    rows = list(csv.DictReader(handle))

times = [float(row['time']) for row in rows]
energy = [float(row['energy']) for row in rows]
enstrophy = [float(row['enstrophy']) for row in rows]
max_velocity = [float(row['max_velocity']) for row in rows]
```

```python
fig, axes = plt.subplots(3, 1, figsize=(8, 9), constrained_layout=True)
axes[0].plot(times, energy, color='#0f766e')
axes[0].set_title('Kinetic Energy')
axes[1].plot(times, enstrophy, color='#b45309')
axes[1].set_title('Enstrophy')
axes[2].plot(times, max_velocity, color='#1d4ed8')
axes[2].set_title('Max Velocity')
for axis in axes:
    axis.set_xlabel('Time')
    axis.grid(alpha=0.25)
plt.show()
```
